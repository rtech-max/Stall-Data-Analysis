```python
import pandas as pd

data = {
    "Stall_Name": [
        "Food Fiesta", "Art Corner", "Game Zone",
        "Craft House", "Tasty Treats", "Photo Booth",
        "Handmade Hub", "Fun Arena", "Sweet Spot",
        "Creative Minds"
    ],
    "Visitors": [250, 150, 300, 180, 280, 220, 130, 270, 200, 160],
    "Units_Sold": [180, 100, 220, 120, 200, 150, 90, 190, 140, 110],
    "Revenue": [9000, 5000, 11000, 6000, 10000, 7500, 4500, 9500, 7000, 5500],
    "Expenses": [5000, 2500, 6000, 3000, 5500, 3500, 2000, 4800, 3500, 2500]
}

df = pd.DataFrame(data)

df["Profit"] = df["Revenue"] - df["Expenses"]

print("STALL DATA")
display(df)

print("TOTAL REVENUE:", df["Revenue"].sum())
print("TOTAL EXPENSES:", df["Expenses"].sum())
print("TOTAL PROFIT:", df["Profit"].sum())
print("TOTAL VISITORS:", df["Visitors"].sum())

print("TOP REVENUE STALL:")
display(df.loc[df["Revenue"].idxmax()])

print("TOP PROFIT STALL:")
display(df.loc[df["Profit"].idxmax()])
```

```python
import matplotlib.pyplot as plt

# Revenue Comparison
plt.figure(figsize=(10, 5))

plt.bar(df["Stall_Name"], df["Revenue"])

plt.title("Stall-wise Revenue Comparison")
plt.xlabel("Stall Name")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(df["Stall_Name"], df["Profit"])

plt.title("Stall-wise Profit Comparison")
plt.xlabel("Stall Name")
plt.ylabel("Profit (₹)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
